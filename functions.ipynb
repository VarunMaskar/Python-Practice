# Functions
A function is a reusable block of code designed to perform a particular task. Functions help to make programs shorter, organized, reusable, and easier to understand.

Why are functions used?
**Code reusability:** Write the code once and use it multiple times.
**Reduces repetition**:Avoid writing the same statements repeatedly.
 **Improves readability:** Divides a large program into smaller parts.
**Makes programs modular**: Different tasks can be handled by different functions.

# Function Names
Function names follow the same rules as variable names in Python:

A function name must start with a letter or underscore
A function name can only contain letters, numbers, and underscores
Function names are case-sensitive (myFunction and myfunction are different)

calculate_sum()
_private_function()
myFunction2()

It's good practice to use descriptive names that explain what the function does.

# Types of Functions in Python
**Built-in functions** – Already provided by Python, such as print(), len(), sum(), and input(). **User-defined functions** – Created by the programmer using the def keyword.
**Functions with parameters** – Accept values as input.
**Functions with return values** – Return a result using the return statement.

Write a Python program to develop a simple Online Shopping Bill System:

The program should accept the details of products purchased by a customer and calculate the final bill as following:

calculate_subtotal
        ↓
calculate_discount
        ↓
Calculate Amount After Discount
        ↓
calculate_tax
        ↓
Calculate Final Amount
        ↓
display_bill

Calculate the discount according to the following rules:
Subtotal ≥ ₹5000 → 20% discount
Subtotal ≥ ₹3000 → 15% discount
Subtotal ≥ ₹1000 → 10% discount
Below ₹1000 → No discount

Accept the amount after discount.
Calculate 18% GST.


display_bill as:

----- ONLINE SHOPPING BILL -----

Subtotal          : ₹4000
Discount          : ₹600
Amount After Disc.: ₹3400
GST (18%)         : ₹612
--------------------------------
Final Amount      : ₹4012
--------------------------------

```python
def calculate_subtotal(price, quantity):
    return price * quantity


def calculate_discount(subtotal):
    if subtotal >= 5000:
        discount = subtotal * 20 / 100
    elif subtotal >= 3000:
        discount = subtotal * 15 / 100
    elif subtotal >= 1000:
        discount = subtotal * 10 / 100
    else:
        discount = 0

    return discount


def calculate_tax(amount):
    return amount * 18 / 100


def display_bill(subtotal, discount, tax, final_amount):
    print("\n----- ONLINE SHOPPING BILL -----")
    print("Subtotal           : ₹", subtotal)
    print("Discount           : ₹", discount)
    print("Amount After Disc. : ₹", subtotal - discount)
    print("GST (18%)          : ₹", tax)
    print("--------------------------------")
    print("Final Amount       : ₹", final_amount)
    print("--------------------------------")


# Main program
price = 4000
quantity = 1

subtotal = calculate_subtotal(price, quantity)

discount = calculate_discount(subtotal)

amount_after_discount = subtotal - discount

tax = calculate_tax(amount_after_discount)

final_amount = amount_after_discount + tax

display_bill(subtotal, discount, tax, final_amount)
```

```python
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```
